```python
from tensorflow.keras.preprocessing.text import one_hot
```

```python
### sentences
sent=[ ' the glass of milk ', ' the glass of juice ', ' the cup of tea ', ' I am a good boy ', ' I am a good developer ', ' understand the meaning of words ' ,'your videos are good ']
```

```python
sent
```

```python
### Define the vocabulary size
vocab_size = 10000
```

#### <font color='sky blue'>Here the senteces are converted into vectors in my 10000 vocabulary size. Each word is represented by a unique integer in the range of 1 to 10000. The one_hot function from Keras is used to achieve this representation.</font>

```python
### OneHot Representation for every word in the sentence
one_hot_rep=[one_hot(words, vocab_size) for words in sent]
one_hot_rep 
```

#### <font color='sky blue'>Imp : We need to make all the sentences of equal size or we will not be able to train RNN.</font>

```python
### Word Embedding Representation

from tensorflow.keras.layers import Embedding
from tensorflow.keras.utils import pad_sequences
from tensorflow.keras.models import Sequential
```

```python
import numpy as np
```

```python
sent_lenght=8
```

```python
embedded_docs=pad_sequences(one_hot_rep, padding='pre', maxlen=sent_lenght) 

## pre = 0's at first and post = 0's at last
```

#### <font color='sky blue'>Representing all the sentences in equal size.</font>

```python
embedded_docs
```

```python
### Feature Representation

dim=10
```

```python
model=Sequential()
model.add(Embedding(vocab_size, dim, input_length=sent_lenght))
model.compile('adam', 'mse')

#### using adam optimizer and mean square error loss function
#### adam optimizer is used for faster convergence and mean square error loss function is used for regression problems
```

```python
model.build(input_shape=(None, sent_lenght))
model.summary()
```

```python
model.predict(embedded_docs)


### Every word in the sentence is represented as a vector of size 10. The embedding layer learns these representations during training.
```
